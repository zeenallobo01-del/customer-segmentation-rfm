```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
Data = {
    'CustomerID': ['C001', 'C002', 'C003', 'C004', 'C005', 'C006', 'C007', 'C008', 'C009', 'C010', 'C011', 'C012'],
    'Recency': [5, 45, 12, 89, 2, 60, 15, 102, 8, 55, 3, 75],
    'Frequency': [12, 2, 8, 1, 15, 3, 10, 1, 14, 2, 18, 2],
    'Monetary': [1450, 210, 950, 75, 1890, 310, 1200, 50, 1650, 250, 2200, 180],
    'CLV': [3200, 450, 2100, 120, 4100, 600, 2800, 90, 3800, 500, 4800, 350],
    'CAC': [150, 120, 100, 90, 180, 110, 130, 95, 160, 105, 200, 115]
}
```

```python
df = pd.DataFrame(Data)
```

```python
scaler = StandardScaler()
```

```python
rfm_features = df[['Recency', 'Frequency', 'Monetary']]
scaled_features = scaler.fit_transform(rfm_features)
```

```python

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(scaled_features)
```

```python
cluster_summary = df.groupby('Cluster')[['Recency', 'Frequency', 'Monetary']].mean()
```

```python
print("Cluster Centers (Averages):\n", cluster_summary)
```

```python
cluster_mapping = {0: 'At-Risk', 1: 'Champions', 2: 'Loyalist'}
df['Segment'] = df['Cluster'].map(cluster_mapping)
```

```python
df['CLV_CAC_Ratio'] = df['CLV']/ df['CAC']
```

```python
print ("\nprocessed Customer Segments:")
print (df[['CustomerID', 'Segment', 'Monetary', 'CLV_CAC_Ratio']])
```
